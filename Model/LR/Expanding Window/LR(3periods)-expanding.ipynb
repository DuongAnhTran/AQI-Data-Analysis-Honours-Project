# Load Data 

```python
# %% Cell 2: Load Data
def loadData():
    df = pd.read_parquet("gafanha.parquet")
    df["timestamp"] = pd.to_datetime(df["timestamp"]).dt.round("h")
    print(df.count())
    return (
        df.pivot_table(index="timestamp", columns="collection", values="value", aggfunc="median")
        .asfreq("h")
        .loc[:"2024-12-31 23:00"]
        .reset_index()
    )

df = loadData()




# %% Cell 3: Readings per day per variable - for validating the dataset only
def checkDailyCount(df):
    df_daily_counts = df.set_index("timestamp").resample("D").count()

    # quick visual: counts per year for each variable
    df_daily_counts.groupby(df_daily_counts.index.year).mean().plot(
        kind="bar", figsize=(12, 5), title="Avg daily reading count per variable, by year"
    )
    plt.ylabel("avg readings/day")
    plt.tight_layout()
    plt.show()
```

```python
# %% Cell 4: Force daily, median and fill in missing values with darts package
from darts.utils.missing_values import fill_missing_values
daily = df.set_index("timestamp").resample("D").median().asfreq("D")
print(daily.count())

# European AQI breakpoints per pollutant
aqi_bins = {
    "PM25": [-np.inf, 10, 20, 25, 50, np.inf],
    "PM10": [-np.inf, 20, 35, 50, 100, np.inf],
    "NO2":  [-np.inf, 40, 100, 200, 400, np.inf],
    "O3":   [-np.inf, 80, 120, 180, 240, np.inf],
    "SO2":  [-np.inf, 100, 200, 350, 500, np.inf],
}
def sub_index(df):
    return pd.concat([pd.cut(df[p], bins=b, labels=[1, 2, 3, 4, 5]).astype(float) for p, b in aqi_bins.items()], axis=1)
def sub_masked(df):  # only count pollutant-days that were really observed
    return sub_index(df).where(daily[pollutants].reindex(df.index).notna())
def aqi_from(df):
    return sub_masked(df).max(axis=1, skipna=True)

# raw observed data only: for the coverage / dominant-pollutant checks, not for modelling
sub_idx = sub_index(daily)
dominant = sub_idx.dropna(how="all").idxmax(axis=1)
print(sub_idx.max(axis=1).last_valid_index())

daily.index = daily.index.tz_localize(None)
series = fill_missing_values(TimeSeries.from_dataframe(daily, freq="D"))
```

# Set up and prepare for training 3 LR Models

***Scaler does have access to the full history of covariates, so thyat might lead to a risk of dataleak, but without it, the LR will crash due to missing values from the variables***

**Has to drop Radiation and Precipitation here because they will contaminate the post period if not. And the importance will be inflated later on (Same for ARIMA. But for RF and XGBoost, they can ignore this)**

```python
# %% Cell 5: targets = 5 pollutants (multivariate); covariates = other variables, scaled
from darts.dataprocessing.transformers import Scaler
from sklearn.preprocessing import StandardScaler

pollutants = list(aqi_bins)
target = series[pollutants]
# NaN -> 0 for covariates, as the scaler cannot handle NaN values
covariates = Scaler(StandardScaler()).fit_transform(
    series.drop_columns(pollutants + ["Radiation", "Precipitation"]).map(lambda x: np.nan_to_num(x)))

# targets + covariates in one series, used by Cell 10
data = target.stack(covariates)   
```

```python
# %% Cell 6: generic per-period pipeline (ONE multivariate model per period - no optuna because LR)
rm = lambda a, b: float(np.sqrt(((a - b) ** 2).mean()))

def aqi_rmse(a, p):
    return rm(aqi_from(p.to_dataframe()), aqi_from(a.slice_intersect(p).to_dataframe()))

def run_period(target, covariates, model_cls, fixed_param):
    y_train, y_temp = target.split_before(0.6)
    y_val, y_test = y_temp.split_before(0.5)
    x_train, x_temp = covariates.split_before(0.6)
    x_val, x_test = x_temp.split_before(0.5)

    y_fit, x_fit = y_train.append(y_val), x_train.append(x_val)
    test_pred = model_cls(**fixed_param).historical_forecasts(
        series=y_fit.append(y_test), past_covariates=x_fit.append(x_test),
        start=y_test.start_time(), forecast_horizon=1, stride=1,
        retrain=True, train_length=None, last_points_only=True, verbose=False,
    )

    # Final model on train+val, used for permutation importance
    model = model_cls(**fixed_param).fit(y_fit, past_covariates=x_fit)
    return dict(model=model, y_test=y_test, test_pred=test_pred,
                sizes=(len(y_train), len(y_val), len(y_test)))
```

# For Each of the Periods

### Config the models - run search

```python
# %% Cell 7: Model Config
import joblib

lr_fixed = dict(lags=7, lags_past_covariates=7, output_chunk_length=1)

# 3 periods: pre, during, post
bounds = {"pre":    (None, "2020-02-29"),
          "during": ("2020-03-01", "2021-04-30"),
          "post":   ("2021-05-01", None)}
```

### Activate the training here

```python
# %% Cell 7b: Run the training and testing for each period
import time
results = {}
for name, (s, e) in bounds.items():
    lo = pd.Timestamp(s) if s else series.start_time()
    hi = pd.Timestamp(e) if e else series.end_time()
    t0 = time.perf_counter()
    r = results[name] = run_period(target.slice(lo, hi), covariates.slice(lo, hi), LinearRegressionModel, lr_fixed)
    print(name, r["sizes"], f"AQI RMSE {aqi_rmse(r['y_test'], r['test_pred']):.3f}", f"{time.perf_counter() - t0:.0f}s")
joblib.dump(results, "results_lr_multi.pkl")
```

# Report, and other analysis visualisation

```python
# %% Cell 9: report
def pair(r):  # (actual, predicted) pollutant frames on the test window
    p = r["test_pred"].to_dataframe()
    return r["y_test"].to_dataframe().loc[p.index], p

def aqi_pair(r):  # returns (actual AQI, predicted AQI) on the test window
    a, p = pair(r)
    return aqi_from(a), aqi_from(p)

# per-pollutant RMSE / MAE
report = pd.concat({n: pd.DataFrame({"rmse": ((p - a) ** 2).mean() ** .5, "mae": (p - a).abs().mean()})
                    for n, r in results.items() for a, p in [pair(r)]})
print(report.round(3))
report.to_csv("report_metrics_lr.csv")

# AQI metrics (RMSE, MAE, MAPE) per period
rows = {}
for n, rp in results.items():
    a, p = aqi_pair(rp)
    e = p - a
    rows[n] = dict(rmse=np.sqrt((e**2).mean()), mae=e.abs().mean(),
                   mape=(e.abs() / a).mean() * 100, n_test=e.notna().sum())
aqi_metrics = pd.DataFrame(rows).T
print(aqi_metrics.round(3))
aqi_metrics.to_excel("aqi_metrics_lr.xlsx")
```

### 1 week and 1 month performance check for each model

The post test window is entirely 2024, when PM10/PM25 are missing, so AQI is built from NO2/O3/SO2 only. The post RMSE at 1, 7 and 30 days therefore measures a different target from pre and during. It isn't comparable across periods, and no code change fixes that. Report post separately or footnote it.

```python
# %% Cell 9b: 1d / 7d / 30d horizon check - use the lambda function rm(a, b) defined in Cell 6 to compute RMSE
def horizon_scores(h):
    cov = covariates.shift(h - 1)
    tgt = target.slice_intersect(cov)
    cov = cov.slice_intersect(tgt)
    out = {}
    for name, (s, e) in bounds.items():
        lo = pd.Timestamp(s) if s else tgt.start_time()
        hi = pd.Timestamp(e) if e else tgt.end_time()
        y, x = tgt.slice(lo, hi), cov.slice(lo, hi)
        y_fit, y_test = y.split_before(0.8)
        x_fit, _ = x.split_before(0.8)
        m = LinearRegressionModel(**lr_fixed).fit(y_fit, past_covariates=x_fit)
        pr = m.historical_forecasts(series=y, past_covariates=x, start=y_test.start_time(),
                                    forecast_horizon=h, stride=1, retrain=False,
                                    last_points_only=True, verbose=False, show_warnings=False)
        a, q = y_test.slice_intersect(pr).to_dataframe(), pr.to_dataframe()
        out.update({(name, p): rm(q[p], a[p]) for p in pollutants})
        out[(name, "AQI")] = rm(aqi_from(q), aqi_from(a))
    return pd.Series(out)

print(pd.DataFrame({f"{h}d": horizon_scores(h) for h in (1, 7, 30)}).round(3))
```

### Variable impact on decision for each period

```python
# %% Cell 10: permutation importance on AQI (built from predicted pollutants) + each pollutant
import warnings, logging
warnings.filterwarnings("ignore"); logging.getLogger("darts").setLevel(logging.ERROR)

# per-model block - change only this block when copying to another model
final_model, model_cls, fixed, lookback = "lr", LinearRegressionModel, lr_fixed, 7
def forecast(model, y, x, start):
    return model.historical_forecasts(series=y, past_covariates=x, start=start, forecast_horizon=1,
        stride=1, retrain=False, last_points_only=True, verbose=False).to_dataframe()

groups = {"PM (PM10+PM25+Particles)": ["PM10", "PM25", "Particles"],
          "NOx family (NO+NO2+NOx)": ["NO", "NO2", "NOx"]}
grouped = {c for g in groups.values() for c in g}
units = {**groups, **{c: [c] for c in data.components if c not in grouped}}

def perm_importance(rp, y_w, n=10, units=units):
    # rp = results[period] (one model); y_w = test window to score on.
    # Returns RMSE increase per variable/group; columns = each pollutant + AQI.
    lo = y_w.start_time() - pd.Timedelta(days=lookback)
    win = data.slice(lo, y_w.end_time())
    truth = win.to_dataframe()
    cols, rng = list(win.components), np.random.default_rng(seed)

    def score(ts):
        pred = forecast(rp["model"], ts[pollutants], ts.drop_columns(pollutants), y_w.start_time())
        act = truth.loc[pred.index]
        return pd.Series({**{p: rm(pred[p], act[p]) for p in pollutants},
                          "AQI": rm(aqi_from(pred), aqi_from(act))})

    base = score(win)
    out = {}
    for u, members in units.items():
        d = []
        for _ in range(n):
            v, i = win.values().copy(), [cols.index(c) for c in members]
            v[:, i] = v[rng.permutation(len(v))][:, i]
            d.append(score(win.with_values(v)) - base)
        out[u] = pd.concat(d, axis=1).mean(axis=1)
    return pd.DataFrame(out).T

def plot_share(df, title):
    ax = df.mul(100).plot.bar(figsize=(12, 5), title=title)
    for t in ax.get_xticklabels():
        if t.get_text() in ("Radiation", "Precipitation"): t.set_color("red")
    for c in ax.containers: ax.bar_label(c, fmt="%.1f", fontsize=7, padding=2)
    ax.set_ylabel("% of total importance (red = data only in 2024)")
    plt.show()

def share(d):
    d = d.clip(lower=0)
    return d / d.sum()

def above_floor(imp):
    return imp.gt(-imp.min().clip(upper=0), axis=1)   # True = importance exceeds the column's largest negative

# each period's own model on its own test window
imp = {k: perm_importance(rp, rp["y_test"]) for k, rp in results.items()}
aqi_imp = pd.DataFrame({k: d["AQI"] for k, d in imp.items()})   # rows = variable, cols = period

plot_share(share(aqi_imp), f"{final_model}: share of AQI error increase when variable is shuffled")
print(aqi_imp.round(5))
print(above_floor(aqi_imp))
# change in share vs pre, in percentage points
print((share(aqi_imp)[["during", "post"]].sub(share(aqi_imp)["pre"], axis=0) * 100).round(1))

pd.concat(imp).to_csv(f"importance_{final_model}.csv")   # all columns (AQI + each pollutant), period x variable
```

### Line graph for models' prediction vs actual data

```python
fig, axes = plt.subplots(3, 1, figsize=(10, 15), sharey=True)
colors = dict(pre="tab:blue", during="tab:orange", post="tab:green")

for ax, (name, rp) in zip(axes, results.items()):
    a, pr = aqi_pair(rp)
    actual_w, pred_w = a.resample("W").mean(), pr.resample("W").mean()

    ax.plot(actual_w.index, actual_w.values, color="black", lw=2, marker="o", markersize=4, label="actual", zorder=2)
    ax.plot(pred_w.index, pred_w.values, color=colors[name], lw=1.2, marker="o", markersize=3,
        linestyle="--", label="predicted (from pollutants)", zorder=3)
    ax.set_title(name)
    ax.tick_params(axis="x", rotation=45)
    ax.legend()

axes[0].set_ylabel("AQI (weekly mean)")
plt.tight_layout()
plt.show()
```

### Daily model prediction vs actual data

```python
fig, axes = plt.subplots(3, 1, figsize=(12, 12), sharey=True)
colors = dict(pre="tab:blue", during="tab:orange", post="tab:green")

for ax, (name, rp) in zip(axes, results.items()):
    actual, pred = aqi_pair(rp)
    rmse_, mae_ = np.sqrt(((pred - actual) ** 2).mean()), (pred - actual).abs().mean()
    mape_ = ((pred - actual).abs() / actual).mean() * 100
    ax.plot(actual.index, actual.values, color="black", lw=1.5, marker="o", markersize=3, label="actual")
    ax.plot(pred.index, pred.values, color=colors[name], lw=1, marker="o", markersize=2,
            linestyle="--", label="predicted (from pollutants)")
    ax.set_title(f"{name}-daily period: {len(actual)} test days, AQI RMSE={rmse_:.2f}, MAE={mae_:.2f}, MAPE={mape_:.2f}%")
    ax.tick_params(axis="x", rotation=45)
    ax.legend()

axes[0].set_ylabel("AQI (daily)")
plt.tight_layout()
plt.show()
```
